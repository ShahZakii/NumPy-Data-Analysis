# Arithmetic Operator
### 1. Addition

```python
import numpy as np

var = np.array([1,2,3,4])
var_add = var + 3

print(var_add)
```

## add( )

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])

var_add1 = np.add(var1, var2)
print(var_add1)
```

### 2. Subtraction

```python
var = np.array([1,2,3,4])
var_sub = var - 3

print(var_sub)
```

## subtract( )

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])

var_sub1 = np.subtract(var1, var2)
print(var_sub1)
```

### 3. Multiplication

```python
var = np.array([1,2,3,4])
var_mul = var * 3

print(var_mul)
```

## multiply( )

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])

var_multi1 = np.multiply(var1, var2)
print(var_multi1)
```

### 4. Divide

```python
var = np.array([1,2,3,4])
var_divide = var / 3

print(var_divide)
```

## divide( )

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])

var_divide1 = np.divide(var1, var2)
print(var_divide1)
```

### 5. Modulus

```python
var = np.array([1,2,3,4])
var_mod = var % 3

print(var_mod)
```

## mod( )


```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])

var_mod1 = np.mod(var1, var2)
print(var_mod1)
```
